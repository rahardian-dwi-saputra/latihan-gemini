```python
import os
import requests
import httpx
from dotenv import load_dotenv
from google import genai
from google.genai import types
from IPython.display import Markdown
```

```python
load_dotenv()

client = genai.Client()

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Explain how AI works"
)

print(response.text)
```

```python
Markdown(response.text)
```

```python
for model in client.models.list():
    print(model.name)
```

```python
response = client.models.generate_content_stream(
    model="gemini-3.5-flash-lite",
    contents="Explain how AI works"
)

# Iterasi setiap potongan teks yang diterima
for chunk in response:
    print(chunk.text, end="", flush=True)
```

```python
image_url = "https://marketplace.canva.com/EAFhLHB9S8g/1/0/900w/canva-pink-putih-minimalist-salon-price-list-28N6AZMRZKI.jpg"

response_http = requests.get(image_url)

mime_type = response_http.headers.get("Content-Type", "image/jpeg")
image_bytes = response_http.content

image_part = types.Part.from_bytes(
    data=image_bytes,
    mime_type=mime_type
)

prompt = "Deskripsikan gambar berikut secara singkat."
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[prompt, image_part]
)

Markdown(response.text)
```

```python
doc_url = "https://discovery.ucl.ac.uk/id/eprint/10089234/1/343019_3_art_0_py4t4l_convrt.pdf"

pdf_bytes = httpx.get(doc_url).content

pdf_part = types.Part.from_bytes(
    data=pdf_bytes,
    mime_type="application/pdf"
)

prompt = "Summarize this document"

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[prompt, pdf_part]
)

print(response.text)
```

```python
url = "https://storage.googleapis.com/generativeai-downloads/data/State_of_the_Union_Address_30_January_1961.mp3"
local_filename = "sample.mp3"

#unduh file audio
response_http = requests.get(url)
with open(local_filename, "wb") as f:
    f.write(response_http.content)

#upload file audio
myfile = client.files.upload(file=local_filename)
print(f"{myfile=}")

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=[myfile, "Describe this audio clip"]
)

print(f"{response.text=}")
```

```python
chat = client.chats.create(
    model="gemini-3.5-flash-lite",
    history=[
        types.Content(
            role="user",
            parts=[types.Part.from_text(text="Hello")]
        ),
        types.Content(
            role="model",
            parts=[types.Part.from_text(text="Great to meet you. What would you like to know?")]
        ),
    ]
)

response = chat.send_message("I have 2 dogs in my house.")
print(response.text)

response2 = chat.send_message("How many paws are in my house?")
print(response2.text)
```
